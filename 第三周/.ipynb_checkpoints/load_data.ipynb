```python
import warnings

import matplotlib.pyplot as plt
# ============================================================
# 0. 导入与全局配置
# ============================================================
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn.datasets import load_breast_cancer
from sklearn.ensemble import RandomForestClassifier
from sklearn.feature_selection import (
    mutual_info_classif, mutual_info_regression,
    SelectKBest, SequentialFeatureSelector
)
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, f1_score
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

warnings.filterwarnings('ignore')
np.random.seed(42)


# ============================================================
# 1. 数据加载
# ============================================================

def load_monk1(train_path="monks-1-clean.train", test_path="monks-1-clean.test"):
    """
    从本地读取 MONK1 数据集。
    UCI 的 monks-1.train / monks-1-clean.test 格式：
      第 1 列：标签（0 或 1）
      第 2-7 列：A1, A2, A3, A4, A5, A6
    """
    train = pd.read_csv(train_path, sep=r'\s+', header=None, engine='python')
    test = pd.read_csv(test_path, sep=r'\s+', header=None, engine='python')
    df = pd.concat([train, test], ignore_index=True).dropna()

    # 数据自检
    print(f"[MONK1] shape={df.shape}, 列数={df.shape[1]}")
    print(f"[MONK1] 第0列唯一值={sorted(df.iloc[:, 0].unique())}")
    assert df.shape[1] == 7, f"MONK1 应有 7 列（标签+6特征），实际 {df.shape[1]} 列"
    assert set(df.iloc[:, 0].unique()) == {0, 1}, "MONK1 标签列应为 0/1"

    X = df.iloc[:, 1:7].astype(int)
    X.columns = [f'A{i}' for i in range(1, 7)]
    y = df.iloc[:, 0].astype(int)
    return X, y


# 乳腺癌数据集
data_bc = load_breast_cancer()
X_bc = pd.DataFrame(data_bc.data, columns=data_bc.feature_names)
y_bc = data_bc.target

# MONK1：加载 + one-hot
X_monk_raw, y_monk = load_monk1()
X_monk = pd.get_dummies(X_monk_raw, columns=X_monk_raw.columns).astype(float)

print("MONK1 raw shape:", X_monk_raw.shape,
      " one-hot shape:", X_monk.shape,
      " 类别分布:", np.bincount(y_monk))
print("Breast Cancer shape:", X_bc.shape,
      " 类别分布:", np.bincount(y_bc))


# ============================================================
# 2. 统一评估协议
# ============================================================

def evaluate_selection(X, y, method_fn, k=10, dataset_name="", downstream="logreg", random_state=42):
    """
    统一评估函数。特征选择只在训练集上做，避免数据泄露。
    """
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=0.3, random_state=random_state, stratify=y
    )

    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)

    # 特征选择（只在训练集上做）
    selected_idx = np.asarray(method_fn(X_train_scaled, y_train, k))

    # 下游模型
    if downstream == "logreg":
        clf = LogisticRegression(max_iter=1000, random_state=random_state)
    else:
        clf = RandomForestClassifier(n_estimators=100, random_state=random_state)

    clf.fit(X_train_scaled[:, selected_idx], y_train)
    y_pred = clf.predict(X_test_scaled[:, selected_idx])

    return {
        "dataset": dataset_name,
        "method": getattr(method_fn, '__name__', 'custom'),
        "downstream": downstream,
        "n_selected": len(selected_idx),
        "accuracy": accuracy_score(y_test, y_pred),
        "f1": f1_score(y_test, y_pred, average='weighted'),
        "selected_indices": list(selected_idx)
    }


# ============================================================
# 3. 四种方法实现
# ============================================================

def mi_method(X_train, y_train, k=10):
    """互信息：单变量过滤法"""
    # 1. 实例化一个“选择器”，告诉它：我要用互信息算分，选出最好的 k 个特征
    selector = SelectKBest(mutual_info_classif, k=k)
    # 2. 喂数据：把训练集的特征 X 和标签 y 给它
    selector.fit(X_train, y_train)
    # 3. 拿结果：返回被选中的特征的“索引”（比如第 0, 3, 5... 列）
    return selector.get_support(indices=True)


def mrmr_method(X_train, y_train, k=10):
    """
    mRMR：最大相关 - 最小冗余（sklearn 无内置）
    """
    #第 1 行：获取特征数
    n_features = X_train.shape[1]

    #第 2 行：算每个特征与标签的互信息（relevance）
    # mutual_info_classif：分类任务用的互信息计算
    #
    # 输入 X_train（398, 30）和 y_train（398,）
    #
    # 输出：一个长度 30 的数组，比如 [0.5, 0.3, 0.8, ...]
    #
    # 每一个数表示：该特征与标签的互信息
    relevance = mutual_info_classif(X_train, y_train, random_state=42)

    # 特征间互信息：连续特征用 mutual_info_regression
    redundancy = np.zeros((n_features, n_features))
    for i in range(n_features):
        for j in range(i + 1, n_features):
            mi = mutual_info_regression(
                X_train[:, [i]], X_train[:, j], random_state=42
            )[0]
            redundancy[i, j] = redundancy[j, i] = mi

    selected = [int(np.argmax(relevance))]
    remaining = set(range(n_features)) - set(selected)

    while len(selected) < k and remaining:
        scores = []
        for i in remaining:
            red = np.mean([redundancy[i, s] for s in selected])
            scores.append(relevance[i] - red)
        best = list(remaining)[int(np.argmax(scores))]
        selected.append(best)
        remaining.remove(best)

    return selected


def forward_method(X_train, y_train, k=10):
    """前向选择：包装法，默认用逻辑回归当筛选器"""
    clf = LogisticRegression(max_iter=1000, random_state=42)
    sfs = SequentialFeatureSelector(
        clf, n_features_to_select=k, direction='forward',
        scoring='accuracy', cv=5, n_jobs=-1
    )
    sfs.fit(X_train, y_train)
    return sfs.get_support(indices=True)


def l1_method_auto(X_train, y_train, C=0.5):
    """
    L1 正则化：嵌入法。不固定特征数，让 C 控制稀疏度。
    """
    clf = LogisticRegression(
        penalty='l1', solver='liblinear', C=C,
        max_iter=1000, random_state=42
    )
    clf.fit(X_train, y_train)
    return np.where(np.abs(clf.coef_[0]) > 1e-6)[0]


def make_forward_with_estimator(estimator, name):
    """生成一个用指定估计器做筛选的前向选择方法"""

    def _method(X_train, y_train, k=10):
        sfs = SequentialFeatureSelector(
            estimator, n_features_to_select=k, direction='forward', scoring='accuracy', cv=5, n_jobs=-1
        )
        sfs.fit(X_train, y_train)
        return sfs.get_support(indices=True)

    _method.__name__ = name
    return _method


# ============================================================
# 4. 运行对比实验
# ============================================================

methods = [mi_method, mrmr_method, forward_method, l1_method_auto]
datasets = [("MONK1", X_monk, y_monk), ("BreastCancer", X_bc, y_bc)]

results = []
for ds_name, X, y in datasets:
    for method in methods:
        res = evaluate_selection(
            X, y, method, k=10, dataset_name=ds_name, downstream="logreg"
        )
        results.append(res)
        print(f"{ds_name:15s} | {method.__name__:18s} | "
              f"n={res['n_selected']:2d} | acc={res['accuracy']:.4f}")

results_df = pd.DataFrame(results)
print("\n=== 结果汇总 ===")
print(results_df[['dataset', 'method', 'n_selected', 'accuracy', 'f1']]
      .to_string(index=False))

# ============================================================
# 5. 可视化
# ============================================================

method_names = [m.__name__ for m in methods]

# 5.1 准确率对比柱状图
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
for ax, ds_name in zip(axes, ["MONK1", "BreastCancer"]):
    subset = results_df[results_df['dataset'] == ds_name] \
        .set_index('method').reindex(method_names)
    x = np.arange(len(method_names))
    ax.bar(x - 0.2, subset['accuracy'], 0.4, label='Accuracy', color='steelblue')
    ax.bar(x + 0.2, subset['f1'], 0.4, label='F1', color='coral')
    ax.set_xticks(x)
    ax.set_xticklabels(method_names, rotation=15)
    ax.set_title(f'{ds_name} — 方法对比')
    ax.set_ylim(0, 1)
    ax.legend()

plt.tight_layout()
plt.show()

# 5.2 特征选择交集热力图
for ds_name in ["MONK1", "BreastCancer"]:
    ds_res = results_df[results_df['dataset'] == ds_name] \
        .set_index('method').reindex(method_names)
    n_m = len(method_names)
    inter = np.zeros((n_m, n_m))
    for i in range(n_m):
        for j in range(n_m):
            s_i = set(ds_res.iloc[i]['selected_indices'])
            s_j = set(ds_res.iloc[j]['selected_indices'])
            inter[i, j] = len(s_i & s_j)

    fig, ax = plt.subplots(figsize=(7, 5.5))
    sns.heatmap(inter, annot=True, fmt='.0f', cmap='Blues',
                xticklabels=method_names, yticklabels=method_names, ax=ax)
    ax.set_title(f'{ds_name} — 四种方法选中特征的交集大小')
    plt.tight_layout()
    plt.show()

# ============================================================
# 6. 下游模型依赖检验
# ============================================================

print("\n=== 下游模型依赖检验 ===")
print("(a) 固定筛选器，换下游模型：")
for ds_name, X, y in datasets:
    res_fwd_log = evaluate_selection(X, y, forward_method, k=10,
                                     dataset_name=ds_name, downstream="logreg")
    res_fwd_rf = evaluate_selection(X, y, forward_method, k=10,
                                    dataset_name=ds_name, downstream="rf")
    res_mi_log = evaluate_selection(X, y, mi_method, k=10,
                                    dataset_name=ds_name, downstream="logreg")
    res_mi_rf = evaluate_selection(X, y, mi_method, k=10,
                                   dataset_name=ds_name, downstream="rf")
    print(f"\n[{ds_name}]")
    print(f"  Forward(LogReg) → LogReg acc: {res_fwd_log['accuracy']:.4f}")
    print(f"  Forward(LogReg) → RF     acc: {res_fwd_rf['accuracy']:.4f}")
    print(f"  MI             → LogReg acc: {res_mi_log['accuracy']:.4f}")
    print(f"  MI             → RF     acc: {res_mi_rf['accuracy']:.4f}")

print("\n(b) 固定下游模型，换筛选器（前向选择'偏心'检验）：")
fwd_logreg = make_forward_with_estimator(
    LogisticRegression(max_iter=1000, random_state=42), "Forward(LogReg)")
fwd_rf = make_forward_with_estimator(
    RandomForestClassifier(n_estimators=100, random_state=42), "Forward(RF)")

for ds_name, X, y in datasets:
    r1 = evaluate_selection(X, y, fwd_logreg, k=10,
                            dataset_name=ds_name, downstream="logreg")
    r2 = evaluate_selection(X, y, fwd_rf, k=10,
                            dataset_name=ds_name, downstream="logreg")
    s1, s2 = set(r1['selected_indices']), set(r2['selected_indices'])
    print(f"\n[{ds_name}]")
    print(f"  Forward(LogReg筛选) 选中: {sorted(s1)}")
    print(f"  Forward(RF筛选)     选中: {sorted(s2)}")
    print(f"  交集大小: {len(s1 & s2)} / 10")
```
